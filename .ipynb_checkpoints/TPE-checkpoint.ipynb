### Unidad de observación y repeticiones

Cada fila de la base corresponde a un registro de ensayo asociado a una configuración de vehículo. Una misma configuración puede aparecer en varios registros debido a diferencias en los procedimientos de prueba, combustibles u otras condiciones del ensayo.
Según los metadatos, la combinación de Veh Mfr Code, Test Vehicle ID y Test Veh Configuration # identifica una configuración de vehículo de ensayo. En consecuencia, la cantidad de filas no equivale necesariamente a la cantidad de configuraciones distintas.
Esta distinción se tendrá en cuenta al identificar posibles duplicados, calcular estadísticas y evaluar las hipótesis. Las repeticiones se examinarán antes de decidir si corresponde conservarlas, seleccionar determinados ensayos o agrupar resultados. Para cada análisis se explicitarán las condiciones consideradas y la unidad sobre la que se obtienen las conclusiones.

```python
# Según los metadatos, estas columnas identifican una configuración.
clave_configuracion = [
    "Veh Mfr Code",
    "Test Vehicle ID",
    "Test Veh Configuration #"
]

# 1. Comparamos la cantidad de registros con las configuraciones distintas.
cantidad_registros = datos.shape[0]

configuraciones = datos[clave_configuracion].drop_duplicates()
cantidad_configuraciones = configuraciones.shape[0]

print("Cantidad de registros:", cantidad_registros)
print("Cantidad de configuraciones distintas:", cantidad_configuraciones)

# 2. Distinguimos filas idénticas de configuraciones repetidas.
# duplicated() considera todas las columnas por defecto.
print(
    "Filas completamente duplicadas:",
    datos.duplicated().sum()
)

# keep=False marca todas las apariciones de una configuración repetida.
configuracion_repetida = datos.duplicated(
    subset=clave_configuracion,
    keep=False
)

cantidad_configuraciones_repetidas = (
    datos.loc[configuracion_repetida, clave_configuracion]
    .drop_duplicates()
    .shape[0]
)

print(
    "Configuraciones presentes en más de un registro:",
    cantidad_configuraciones_repetidas
)

# 3. Mostramos un ejemplo concreto presente en esta base.
# Filtramos por la clave completa para seleccionar una sola configuración.
ejemplo = datos[
    (datos["Veh Mfr Code"] == "ASX")
    & (datos["Test Vehicle ID"] == "5723PT5601")
    & (datos["Test Veh Configuration #"] == 0)
]

columnas_comparacion = clave_configuracion + [
    "Represented Test Veh Model",
    "Test Number",
    "Test Procedure Description",
    "Test Category",
    "Test Fuel Type Description",
    "RND_ADJ_FE"
]

display(
    ejemplo[columnas_comparacion]
    .sort_values(by=["Test Category", "Test Number"])
)
```

La cantidad de registros supera la cantidad de configuraciones distintas, lo que muestra que algunas configuraciones aparecen en múltiples filas. En el ejemplo presentado, una misma configuración tiene registros correspondientes a las categorías FTP y HWY, con diferentes resultados de economía de combustible

### Valores Nulos

Se analiza la presencia de valores faltantes para evaluar la disponibilidad de información en cada atributo. Se calcula su cantidad y su porcentaje respecto del total de registros, y luego se examinan algunos patrones según las características del ensayo.
Esta distinción es necesaria porque un valor ausente puede corresponder a información no registrada o a un atributo que no resulta aplicable a determinado ensayo. En esta etapa se identifican y describen los faltantes; su tratamiento se abordará durante la limpieza.

```python
# 1. Valores faltantes reconocidos por Pandas.
cantidad_nulos = datos.isna().sum()

resumen_nulos = pd.DataFrame({
    "Cantidad de nulos": cantidad_nulos,
    "Porcentaje de nulos": cantidad_nulos / datos.shape[0] * 100
}).sort_values(by="Cantidad de nulos", ascending=False)

display(
    resumen_nulos[resumen_nulos["Cantidad de nulos"] > 0].round(2)
)

print("Atributos con nulos:", (cantidad_nulos > 0).sum())
print(
    "Filas con al menos un nulo:",
    (datos.isna().sum(axis=1) > 0).sum()
)
print("Total de filas:", datos.shape[0])


# 2. Posibles representaciones alternativas de valores faltantes.
# Se ignoran las diferencias de mayúsculas y los espacios exteriores.
marcadores_texto = [
    "", "-", "--", "?", "na", "n/a", "nan", "none", "null",
    "unknown", "missing", "not available", "not applicable", "-1"
]

posibles_faltantes = {}

for columna in datos.columns:
    valores = datos[columna]

    if valores.dtype == "object":
        valores = valores.str.strip().str.lower()
        cantidad = valores.isin(marcadores_texto).sum()
    else:
        cantidad = (valores == -1).sum()

    posibles_faltantes[columna] = cantidad

resumen_marcadores = pd.DataFrame.from_dict(
    posibles_faltantes,
    orient="index",
    columns=["Posibles marcadores de faltantes"]
)

resumen_marcadores = resumen_marcadores[
    resumen_marcadores["Posibles marcadores de faltantes"] > 0
].sort_values(by="Posibles marcadores de faltantes", ascending=False)

display(resumen_marcadores)
```

### Examinar un caso donde el contexto explica los faltantes: ADFE

Los cuatro campos ADFE se relacionan con la economía de combustible derivada analíticamente. Podemos contrastar su disponibilidad con el indicador Analytically Derived FE?

```python
columnas_adfe = [
    "ADFE Test Number",
    "ADFE Total Road Load HP",
    "ADFE Equiv. Test Weight (lbs.)",
    "ADFE N/V Ratio"
]

for respuesta in datos["Analytically Derived FE?"].unique():
    grupo = datos[
        datos["Analytically Derived FE?"] == respuesta
    ]

    print("Analytically Derived FE?:", respuesta)
    print("Cantidad de registros:", grupo.shape[0])

    display(grupo[columnas_adfe].isna().sum())
```

La ausencia de información en los campos ADFE coincide con los registros que no presentan economía de combustible derivada analíticamente. Este patrón es consistente con atributos que no resultan aplicables a esos registros, en lugar de una pérdida general de información. Por ello, su elevado porcentaje de faltantes no es suficiente, por sí solo, para justificar su eliminación o imputación.

Se encontraron valores faltantes en 24 de los 67 atributos. Su proporción varía entre columnas: por ejemplo, ADFE Test Number presenta un 95,05 % de faltantes, mientras que RND_ADJ_FE presenta un 5,16 %.

Se cuantificaron los valores faltantes reconocidos por Pandas y se buscaron posibles representaciones alternativas de ausencia de información. Los valores encontrados en esta segunda revisión se consideran candidatos a inspeccionar, sin clasificarlos automáticamente como nulos ni modificar la base, ya que su interpretación depende del significado de cada atributo.

### Test Veh Displacement (L)

```python
valores = datos["Test Veh Displacement (L)"].dropna()

q1 = valores.quantile(0.25)
q3 = valores.quantile(0.75)
iqr = q3 - q1

limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

print("Límite inferior:", limite_inferior)
print("Límite superior:", limite_superior)

columnas = [
    "Represented Test Veh Make",
    "Represented Test Veh Model",
    "Test Fuel Type Description",
    "Test Veh Displacement (L)"
]

outliers_inf = datos[
    datos["Test Veh Displacement (L)"] < limite_inferior
].sort_values("Test Veh Displacement (L)")

outliers_sup = datos[
    datos["Test Veh Displacement (L)"] > limite_superior
].sort_values("Test Veh Displacement (L)")

print("Cantidad de outliers inferiores:", outliers_inf.shape[0])
display(outliers_inf[columnas])

print("Cantidad de outliers superiores:", outliers_sup.shape[0])
display(outliers_sup[columnas])
```

Vemos que la mayoria de los valores atipicos son a causa de autos electricos, en el contexto de la variable que mide la cilindrada los vehiculos electricos no deberian ser tenidos en cuenta por no tener cilindros. Vamos a asignarle NaN a las filas que correspondan a estos vehiculos:

```python
mascara = datos_limpios["Test Fuel Type Description"] == "Electricity"
print("Filas afectadas:", mascara.sum())

datos_limpios.loc[mascara, "Test Veh Displacement (L)"] = np.nan
```

Ahora verificamos si quedan valores atipicos correspondientes a vehiculos a combustion.

```python
valores = datos_limpios["Test Veh Displacement (L)"].dropna()

q1 = valores.quantile(0.25)
q3 = valores.quantile(0.75)
iqr = q3 - q1

limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

print("Límite inferior:", limite_inferior)
print("Límite superior:", limite_superior)

columnas = [
    "Represented Test Veh Make",
    "Represented Test Veh Model",
    "Test Fuel Type Description",
    "Test Veh Displacement (L)"
]

outliers_inf = datos_limpios[
    datos_limpios["Test Veh Displacement (L)"] < limite_inferior
].sort_values("Test Veh Displacement (L)")

outliers_sup = datos_limpios[
    datos_limpios["Test Veh Displacement (L)"] > limite_superior
].sort_values("Test Veh Displacement (L)")

print("Cantidad de outliers inferiores:", outliers_inf.shape[0])
display(outliers_inf[columnas])

print("Cantidad de outliers superiores:", outliers_sup.shape[0])
display(outliers_sup[columnas])
```

```python
combustible = datos["Test Fuel Type Description"]
frecuencias_combustible = combustible.value_counts()

display(
    frecuencias_combustible.rename("Cantidad")
    .to_frame()
    .assign(
        Porcentaje=(
            frecuencias_combustible / combustible.notna().sum() * 100
        ).round(2)
    )
)

print("Valores faltantes:", combustible.isna().sum())
print("Etiquetas vacías:", combustible.dropna().str.strip().eq("").sum())
print(
    "Etiquetas con espacios al inicio o al final:",
    combustible.dropna().ne(combustible.dropna().str.strip()).sum()
)
print("Categorías originales:", combustible.nunique())
print(
    "Categorías al normalizar espacios y mayúsculas:",
    combustible.dropna().str.strip().str.casefold().nunique()
)
```

Encontramos 11 categorías, coincidiendo con la cantidad indicada en la metadata. No hay valores faltantes, etiquetas vacías ni categorías duplicadas por diferencias de mayúsculas o espacios.

Tier 2 Cert Gasoline concentra 3.236 registros (75,48 %), seguida de Electricity, con 654 (15,26 %). Revisamos las seis categorías con menos de 20 registros para evaluar si su baja frecuencia está acompañada de alguna inconsistencia.

```python
# Seleccionamos categorías poco frecuentes para revisión.
# Este umbral no indica que sean errores.
categorias_poco_frecuentes = frecuencias_combustible[
    frecuencias_combustible < 20
].index

filas_combustible = datos[
    datos["Test Fuel Type Description"].isin(
        categorias_poco_frecuentes
    )
]

agrupado_combustible = (
    filas_combustible.groupby(
        [
            "Test Fuel Type Description",
            "Represented Test Veh Make",
            "Represented Test Veh Model",
            "Test Procedure Description"
        ],
        dropna=False
    )
    .size()
    .reset_index(name="Cantidad")
)

with pd.option_context(
    "display.max_rows", None,
    "display.max_columns", None
):
    display(agrupado_combustible)
```

Hydrogen 5 aparece en diez registros de los modelos Honda CR-V e:FCEV, Hyundai NEXO y Toyota MIRAI. Las categorías de gasolina menos frecuentes aparecen en ensayos de Lotus Emira y Pagani Utopia. E85 tiene 12 registros.

La baja frecuencia de estos combustibles no demuestra un error de carga. Las categorías permiten distinguir combustibles y especificaciones de ensayo, por lo que no corresponde agruparlas únicamente por contener términos similares.

```python
correspondencia_combustible = (
    datos.groupby(
        ["Test Fuel Type Cd", "Test Fuel Type Description"],
        dropna=False
    )
    .size()
    .reset_index(name="Cantidad")
)

display(correspondencia_combustible)
```

Cada código de combustible corresponde a una única descripción y cada descripción a un único código. No encontramos inconsistencias internas que justifiquen reemplazar categorías por NaN.

Conservamos las 11 categorías, incluidas las menos frecuentes. También mantenemos separadas las distintas especificaciones de gasolina, ya que tienen códigos propios y representan categorías diferentes dentro del archivo.

### CO2 (g/mi)

#### Análisis de valores atípicos de CO2

Analizamos los valores atípicos utilizando el criterio de 1,5 veces el rango intercuartílico. También revisamos los valores iguales a cero, ya que pueden ser coherentes para ensayos con electricidad, pero requieren una revisión en otros combustibles.

```python
co2 = datos["CO2 (g/mi)"].dropna()

q1_co2 = co2.quantile(0.25)
q3_co2 = co2.quantile(0.75)
iqr_co2 = q3_co2 - q1_co2

limite_inferior_co2 = q1_co2 - 1.5 * iqr_co2
limite_superior_co2 = q3_co2 + 1.5 * iqr_co2

print("Límite inferior:", round(limite_inferior_co2, 2))
print("Límite superior:", round(limite_superior_co2, 2))
print("Atípicos inferiores:", (co2 < limite_inferior_co2).sum())
print("Atípicos superiores:", (co2 > limite_superior_co2).sum())

ceros_co2 = datos[datos["CO2 (g/mi)"] == 0]

display(
    ceros_co2.groupby("Test Fuel Type Description", dropna=False)
    .size()
    .reset_index(name="Cantidad")
)
```

No hay valores atípicos inferiores según el criterio del IQR. Encontramos 71 registros con CO2 igual a cero: 70 corresponden a electricidad y uno a Cold CO Premium (Tier 2). Revisamos este último caso para determinar si presenta una inconsistencia.

```python
ceros_revisar_co2 = datos[
    (datos["CO2 (g/mi)"] == 0)
    & (datos["Test Fuel Type Description"] != "Electricity")
]

columnas_revision = [
    "Test Number",
    "Veh Mfr Code",
    "Test Vehicle ID",
    "Test Veh Configuration #",
    "Represented Test Veh Model",
    "Test Fuel Type Description",
    "Test Procedure Description",
    "CO2 (g/mi)",
    "CO (g/mi)",
    "THC (g/mi)",
    "RND_ADJ_FE"
]

with pd.option_context("display.max_columns", None):
    display(ceros_revisar_co2[columnas_revision])

# Buscamos otros ensayos de la misma configuración.
claves = [
    "Veh Mfr Code",
    "Test Vehicle ID",
    "Test Veh Configuration #"
]

comparacion_co2 = datos.merge(
    ceros_revisar_co2[claves].drop_duplicates(),
    on=claves,
    how="inner"
)

with pd.option_context(
    "display.max_columns", None,
    "display.max_rows", None
):
    display(comparacion_co2[columnas_revision])
```

El ensayo LTYX10056690, correspondiente al modelo RC F, registra CO2 igual a cero utilizando combustible Cold CO Premium (Tier 2). El mismo ensayo presenta emisiones de CO y THC, y otros ensayos de la misma configuración registran CO2 positivo.

Consideramos este cero como un dato sospechoso y lo reemplazamos por NaN para el análisis, sin asignarle un valor estimado. Conservamos los ceros correspondientes a electricidad.

```python
# Conservamos las correcciones anteriores si ya existe datos_limpios.
if "datos_limpios" not in globals():
    datos_limpios = datos.copy()

caso_sospechoso_co2 = (
    (datos_limpios["Test Number"] == "LTYX10056690")
    & (datos_limpios["CO2 (g/mi)"] == 0)
)

datos_limpios.loc[
    caso_sospechoso_co2, "CO2 (g/mi)"
] = np.nan
```

Ahora revisamos los valores atípicos superiores. En el archivo original hay 68 registros por encima de 617,93 g/mi, equivalentes al 1,91 % de los datos disponibles. Los agrupamos por modelo y procedimiento para facilitar su revisión.

```python
outliers_co2 = datos[
    datos["CO2 (g/mi)"] > limite_superior_co2
]

agrupado_co2 = (
    outliers_co2.groupby(
        [
            "Represented Test Veh Make",
            "Represented Test Veh Model",
            "Test Procedure Description"
        ],
        dropna=False
    )["CO2 (g/mi)"]
    .agg(
        Cantidad="size",
        Minimo="min",
        Maximo="max"
    )
    .reset_index()
    .sort_values("Maximo", ascending=False)
)

with pd.option_context(
    "display.max_rows", None,
    "display.max_columns", None
):
    display(agrupado_co2.round(2))
```

Los valores más elevados corresponden a modelos como BMW M5 Sedan, M5 Touring, XM y Bugatti Pur Sport. El máximo es de 1.093,64 g/mi, registrado para un M5 Sedan en el procedimiento SC03.

Los valores elevados aparecen en distintos modelos y procedimientos. Con la información disponible no podemos clasificarlos como errores de carga, por lo que los conservamos como valores atípicos estadísticos.

### NOx (g/mi)

#### Análisis de valores atípicos de NOx

Analizamos los valores atípicos utilizando el criterio de 1,5 veces el rango intercuartílico. Revisamos los valores iguales a cero y luego los registros elevados según el combustible y el procedimiento de ensayo.

```python
nox = datos["NOx (g/mi)"].dropna()

q1_nox = nox.quantile(0.25)
q3_nox = nox.quantile(0.75)
iqr_nox = q3_nox - q1_nox

limite_inferior_nox = q1_nox - 1.5 * iqr_nox
limite_superior_nox = q3_nox + 1.5 * iqr_nox

print("Límite inferior:", round(limite_inferior_nox, 6))
print("Límite superior:", round(limite_superior_nox, 6))
print("Atípicos inferiores:", (nox < limite_inferior_nox).sum())
print("Atípicos superiores:", (nox > limite_superior_nox).sum())
print("Valores negativos:", (nox < 0).sum())

ceros_nox = datos[datos["NOx (g/mi)"] == 0]

display(
    ceros_nox.groupby("Test Fuel Type Description", dropna=False)
    .size()
    .reset_index(name="Cantidad")
)
```

No hay valores negativos ni atípicos inferiores. Encontramos 129 registros iguales a cero: 90 corresponden a electricidad y 39 a Tier 2 Cert Gasoline.

Los ceros en gasolina requieren considerar la precisión y el método de medición. El archivo no permite determinar si corresponden a redondeo, mediciones bajo el límite de detección u otra causa, por lo que no los reemplazamos automáticamente.

```python
outliers_nox = datos[
    datos["NOx (g/mi)"] > limite_superior_nox
]

resumen_procedimiento_nox = (
    outliers_nox.groupby(
        "Test Procedure Description",
        dropna=False
    )["NOx (g/mi)"]
    .agg(
        Cantidad="size",
        Minimo="min",
        Maximo="max"
    )
    .reset_index()
    .sort_values("Cantidad", ascending=False)
)

resumen_combustible_nox = (
    outliers_nox.groupby(
        "Test Fuel Type Description",
        dropna=False
    )
    .size()
    .reset_index(name="Cantidad")
    .sort_values("Cantidad", ascending=False)
)

display(resumen_procedimiento_nox.round(6))
display(resumen_combustible_nox)
```

Encontramos 198 valores superiores a 0,027285 g/mi, equivalentes al 5,82 % de los registros disponibles. De ellos, 99 corresponden al procedimiento US06 y 35 a Cold CO.

Los valores atípicos aparecen en distintos combustibles, incluidos gasolina y diésel. Revisamos los modelos y las mediciones más elevadas antes de decidir su tratamiento.

```python
agrupado_nox = (
    outliers_nox.groupby(
        [
            "Represented Test Veh Make",
            "Represented Test Veh Model",
            "Test Fuel Type Description",
            "Test Procedure Description"
        ],
        dropna=False
    )["NOx (g/mi)"]
    .agg(
        Cantidad="size",
        Minimo="min",
        Maximo="max"
    )
    .reset_index()
    .sort_values("Maximo", ascending=False)
)

with pd.option_context(
    "display.max_rows", None,
    "display.max_columns", None
):
    display(agrupado_nox.round(6))
```

El valor máximo es de 0,1058 g/mi y corresponde a un Hyundai Venue en el procedimiento US06. También aparecen registros elevados de modelos como GMC Sierra 4WD AT4X y Hyundai Elantra N.

No encontramos una inconsistencia que justifique reemplazar estas mediciones por NaN. Conservamos los valores atípicos, ya que superar el límite del IQR no demuestra un error de carga ni implica superar un límite normativo de emisiones.

### RND_ADJ_FE:

```python
outliers_rnd = datos_limpios[datos_limpios["RND_ADJ_FE"] >= 72.7]

outliers_rnd = (
    outliers_rnd.dropna(subset=["RND_ADJ_FE"])
         .sort_values(by="RND_ADJ_FE", ascending=True)
)

with pd.option_context("display.max_columns", None):
    display(outliers_rnd)
```

Vemos que hay muchos outliers, no podemos mirarlos todos, ordenamos de manera ascendente y vemos que pasa con los ultimos 50:

```python
with pd.option_context("display.max_columns", None):
    display(outliers_rnd.tail(50))
```

Aca notamos que el ultimo valor coherente es el 207.0, a partir de ahi los valores son 999.0 y superiores, lo que no representa una economia de combustible posible. Por ende, decidimos asignarles NaN.

```python
datos_limpios.loc[datos_limpios["RND_ADJ_FE"] >= 999.0, "RND_ADJ_FE"] = np.nan
```
